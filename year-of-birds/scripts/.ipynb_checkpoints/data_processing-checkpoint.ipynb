```python
import pandas as pd
```

```python
df = pd.read_csv("../raw/data.csv")
df
```

```python
# Remove unnecessary columns
df = df[
    [
        "Submission ID",
        "Common Name",
        "Count",
        "State/Province",
        "County",
        "Location",
        "Latitude",
        "Longitude",
        "Date",
        "Time",
        "ML Catalog Numbers",
    ]
]
df
```

```python
# Filter to just this year
df["Date"] = pd.to_datetime(df["Date"])
df = df[df["Date"].dt.year == 2026]
df
```

```python
df = df[
    ~df["Common Name"].str.contains(" sp.")
    & ~df["Common Name"].str.contains("/")
    & ~df["Common Name"].str.contains("hybrid")
    & ~df["Common Name"].str.contains("Domestic")
]
```

```python
df["ML Catalog Numbers"] = df["ML Catalog Numbers"].fillna("")
df["Count"] = df["Count"].replace("X", 1)
df["Count"] = df["Count"].astype(int)
df
```

```python
# Generate table for state lists
# State, Bird, Order
df_results = pd.DataFrame([], columns=["State/Province", "Common Name", "Position"])
for state in df["State/Province"].unique():
    df_state = df[df["State/Province"] == state]
    df_state = df_state.sort_values(["Date", "Time"])
    df_state = df_state[["State/Province", "Common Name"]]
    df_state = df_state.drop_duplicates(keep="first")
    df_state = df_state.reset_index(drop=True)
    df_state["Position"] = df_state.index + 1
    df_results = pd.concat([df_results, df_state], ignore_index=True)

df_results = df_results.rename(
    columns={"State/Province": "state", "Common Name": "name", "Position": "position"}
)
df_results.to_csv("../data/lists.csv", index=False)
df_results
```

```python
# Generate table for media
# State, Bird, Media
df_media = df.groupby(["State/Province", "Common Name"], as_index=False).agg(
    {"ML Catalog Numbers": " ".join}
)
df_media["ML Catalog Numbers"] = (
    df_media["ML Catalog Numbers"].str.replace(r"\s+", " ", regex=True).str.strip()
)
df_media = df_media.rename(
    columns={
        "State/Province": "state",
        "Common Name": "name",
        "ML Catalog Numbers": "media",
    }
)
df_media.to_csv("../data/media.csv", index=False)
df_media
```

```python
# Generate table for county based observations
# State, County, Bird, Observations, Count
df_county = df.groupby(["State/Province", "County", "Common Name"], as_index=False).agg(
    observations=("Count", "count"), count=("Count", "sum")
)
df_county = df_county.rename(
    columns={
        "State/Province": "state",
        "Common Name": "name",
        "County": "county",
    }
)
df_county.to_csv("../data/county.csv", index=False)
df_county
```

```python
# Generate table for date based observations
# State, Date, Bird, Count
df_date = df.groupby(["State/Province", "Date", "Common Name"], as_index=False).agg(
    count=("Count", "sum")
)
df_date = df_date.rename(
    columns={
        "State/Province": "state",
        "Common Name": "name",
        "Date": "date",
    }
)
df_date.to_csv("../data/date.csv", index=False)
df_date
```
