```python
import cmath
import math
from pathlib import Path

import matplotlib.image as mpimg
import matplotlib.patches as patches
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_csv("data/king_barchart.csv", index_col=[0])
df
```

```python
df = df.loc[birds.keys()]
df
```

```python
birds = {
    # "Common Goldeneye": ("goldeneye.png", "#654549"),
    "Dark-eyed Junco": ("junco.png", "#916B53"),
    "Common Merganser": ("merganser.png", "#365d37"),
    "Anna's Hummingbird": ("hummingbird.png", "#C91478"),
    # "European Starling": ("starling.png", "#4D3F58"),
    "Cedar Waxwing": ("waxwing.png", "#D4A780"),
    "Osprey": ("osprey.png", "#1B1211"),
    "skipped 1": None,
    "skipped 2": None,
    "Spotted Towhee": ("towhee.png", "#bf5329"),
    "Northern Flicker": ("flicker.png", "#c91f13"),
    "Golden-crowned Kinglet": ("kinglet.png", "#f1dd58"),
    "Bufflehead": ("bufflehead.png", "#1C395B"),
    "Glaucous-winged Gull": ("gull.png", "#b2bac5"),
}
```

```python
R1 = 2
R2 = 4
BACKGROUND = "#f5f1e6"
FONT_PATH = Path("font/ShadowsIntoLight-Regular.ttf")

plt.rcParams["figure.figsize"] = (12, 8)
plt.figure(dpi=400, facecolor=BACKGROUND)
fig, axs = plt.subplots(3, 4)
plt.subplots_adjust(wspace=0, hspace=0)
fig.patch.set_facecolor(BACKGROUND)
for i, bird in enumerate(birds.keys()):
    ax = axs[int(i / 4), i % 4]
    if i == 5 or i == 6:
        ax.axis("off")
        continue

    file, color = birds[bird]

    # Bird name
    ax.text(0, R2 - 0.3, bird, ha="center", font=FONT_PATH, fontsize=16)

    # Abundance chart
    for i, value in enumerate(df.loc[bird]):
        theta = i * 2 * math.pi / len(df.columns)
        r1 = R1
        r2 = R1 + (R2 - R1) * value
        c1 = cmath.rect(r1, theta)
        c2 = cmath.rect(r2, theta)
        ax.plot(
            [c1.real, c2.real],
            [c1.imag, c2.imag],
            lw=4,
            color=color,
            solid_capstyle="round",
        )

    size = R2 - 0.2
    ax.set_xlim([-size, size])
    ax.set_ylim([-size, size + 0.2])

    # Render bird image and crop it
    img = mpimg.imread("photos/" + file)
    im = ax.imshow(img, zorder=15, extent=[-R1, R1, -R1, R1])
    patch = patches.Circle((0, 0), radius=R1, transform=ax.transData)
    im.set_clip_path(patch)

    ax.axis("off")

axs[1, 1].text(1, 0.5, "Birds of Seattle", font=FONT_PATH, fontsize=48, ha="center")

plt.savefig("postcard.png", dpi=600, bbox_inches="tight")
```
