```python
import pandas as pd
```

```python
region = "union_bay"
```

```python
df = pd.read_csv("data/" + region + "_barchart.tsv", sep="\t", header=None, index_col=0)
df = df.drop(columns=[49])
df.index.name = None
df
```

```python
# Filter out hybrids and generics
df = df[
    [
        ("sp." not in bird) and (" x " not in bird) and ("/" not in bird)
        for bird in df.index
    ]
].copy()
df
```

```python
df.to_csv("data/" + region + "_barchart.csv")
```
